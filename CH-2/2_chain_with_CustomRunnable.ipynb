```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI



load_dotenv()


if os.environ.get("OPENAI_API_KEY") :
    print("Bro API key variable is found")

else :
    raise ValueError("OPENAI_APKI_KEY not found")

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
```

# **Chain with Custom Runnable**

```python
#TASK-1 Prompt
prompt_template = ChatPromptTemplate(
    [
    ("system", "You are a helpful assistant"),
    ("human","{input}")
    ]
)
```

```python
#Task -2 LLM
llm_openai = ChatOpenAI(model="gpt-5-mini",temperature=0)
```

```python
# Task - 3 String parser
str_parser = StrOutputParser()
```

```python
# Task - 4 Custom Runnable
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text : str) -> dict :
    
    return({"text" : text})

dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

```python
#Task - 5 Template for Post

prompt_post = ChatPromptTemplate.from_messages(
    messages=[
    ("system", "You are a social media post generator"),
    ("human","Create a promp for the following text for linkedIn : {text}")
    ]
)
```

```python
# Task - 6 Generate the post so call LLM

# Task - 7 String Parser

#will not define them agin as they are defined above
```

```python
# Chain 

chain = prompt_template| llm_openai | str_parser | dictionary_maker_runnable | prompt_post | llm_openai | str_parser

chain.invoke("AI will not take jobs")
```
