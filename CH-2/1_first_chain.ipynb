# **Chains in Langchain**

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI



load_dotenv()


if os.environ.get("OPENAI_API_KEY") :
    print("Bro API key variable is found")

else :
    raise ValueError("OPENAI_APKI_KEY not found")

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
```

## **First chain**

```python

#TASK-1 Prompt
prompt_template = ChatPromptTemplate(
    [
    ("system", "You are a helpful assistant"),
    ("human","{input}")
    ]
)
```

```python
#Task -2 LLM

llm_openai = ChatOpenAI(model="gpt-5-mini",temperature=0)
```

```python
# Task - 3 String parser
str_parser = StrOutputParser()
```

### **Manual Invocation**

```python
template = prompt_template.invoke({"input" : "What is the capital for Spain?"})

result = llm_openai.invoke(template).content
result
```

### **Chain Invocation**

```python
chain =  prompt_template | llm_openai | str_parser

chain.invoke({"input" : "What is the capital of france?"})
```
