```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os

load_dotenv()


if os.environ.get("OPENAI_API_KEY") :
    print("Bro API key variable is found")

else :
    raise ValueError("OPENAI_APKI_KEY not found")
```

```python

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
my_messages = [
    SystemMessage(content= "You are a Gen-Z assistant who always answers in a fun way"),
    HumanMessage(content =  "Bro tell me a fact")
]

llm_openai.invoke(my_messages).content
```
