# **MESSAGES**

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os

load_dotenv()


if os.environ.get("OPENAI_API_KEY") :
    print("Bro API key variable is found")

else :
    raise ValueError("OPENAI_APKI_KEY not found")
```

```python

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)
my_messages = [
    SystemMessage(content= "You are a Gen-Z assistant who always answers in a fun way"),
    HumanMessage(content =  "Bro tell me a fact")
]

llm_openai.invoke(my_messages).content
```

# **PROMPTS**

Prompts are the messages that are sent to the LLM. Prompts are more user-friendly and dynamic than messages.


```python
from langchain_core.prompts import PromptTemplate

topic = input("Enter a topic for fun fact : ")
dynamic_prompt =  PromptTemplate.from_template("Write a fun fact about {topic}")

ready_prompt =  dynamic_prompt.invoke({"topic" : topic})

llm_openai.invoke(ready_prompt).content
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages ([
    ("system", "You are a polite assistant"),
    ("user","Write a fun fact about {topic}")
])

user_input = input("enter the topic")
ready_prompt =  prompt_template.invoke({"topic" : user_input})

llm_openai.invoke(ready_prompt.messages).content
```
