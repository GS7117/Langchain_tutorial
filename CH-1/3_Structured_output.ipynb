```python
from langchain_core.prompts import PromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(model="gpt-5-mini", temperature=0)

result = llm_openai.invoke("Tell me a joke. Generate the output in key:Value pair format whith following keys : setup,punchline")
result.content
```

# **Using Pydantic Model**

```python
from pydantic import BaseModel, Field

class llm_schema(BaseModel) :
     setup: str = Field(description="Setup for the joke") 
     punchline : str = Field(description="Punchline for the joke")
     
```

```python
obj = llm_schema(**{"setup" : "Some Play" , "punchline" : "Some punchline"} )
obj
```

```python
llm_structured_output = llm_openai.with_structured_output(llm_schema)

llm_structured_output.invoke("Tell me a 2 liner motivational quote")
```

# **Uing TypeDict**

```python
from typing import TypedDict

class llm_schema_td(TypedDict) : 
    setup : str
    punchline : str
    
```

```python
llm_structured_output_td = llm_openai.with_structured_output(llm_schema_td)

result = llm_structured_output_td.invoke("Tell me a joke")

result #gives output in dictionary and not an object, also fixes typecasting errors
```
